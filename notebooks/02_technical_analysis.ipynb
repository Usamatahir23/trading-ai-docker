# Technical Analysis

This notebook covers technical indicators and their use in trading strategies.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf
import ta

print("Libraries imported successfully!")
```

## Load Data

```python
symbol = "AAPL"
ticker = yf.Ticker(symbol)
data = ticker.history(period="1y")

print(f"Loaded {len(data)} days of data")
```

## Moving Averages

```python
# Calculate Simple Moving Averages
data['SMA_20'] = ta.trend.SMAIndicator(data['Close'], window=20).sma_indicator()
data['SMA_50'] = ta.trend.SMAIndicator(data['Close'], window=50).sma_indicator()
data['EMA_12'] = ta.trend.EMAIndicator(data['Close'], window=12).ema_indicator()
data['EMA_26'] = ta.trend.EMAIndicator(data['Close'], window=26).ema_indicator()

# Plot
plt.figure(figsize=(14, 8))
plt.plot(data.index, data['Close'], label='Close Price', linewidth=2)
plt.plot(data.index, data['SMA_20'], label='SMA 20', alpha=0.7)
plt.plot(data.index, data['SMA_50'], label='SMA 50', alpha=0.7)
plt.title(f'{symbol} Price with Moving Averages')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## RSI (Relative Strength Index)

```python
# Calculate RSI
data['RSI'] = ta.momentum.RSIIndicator(data['Close'], window=14).rsi()

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

ax1.plot(data.index, data['Close'], label='Close Price', linewidth=2)
ax1.set_ylabel('Price ($)')
ax1.set_title(f'{symbol} Price and RSI')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(data.index, data['RSI'], label='RSI', color='purple', linewidth=2)
ax2.axhline(y=70, color='r', linestyle='--', label='Overbought (70)')
ax2.axhline(y=30, color='g', linestyle='--', label='Oversold (30)')
ax2.set_ylabel('RSI')
ax2.set_xlabel('Date')
ax2.set_ylim(0, 100)
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## MACD (Moving Average Convergence Divergence)

```python
# Calculate MACD
macd = ta.trend.MACD(data['Close'])
data['MACD'] = macd.macd()
data['MACD_Signal'] = macd.macd_signal()
data['MACD_Diff'] = macd.macd_diff()

# Plot
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

ax1.plot(data.index, data['Close'], label='Close Price', linewidth=2)
ax1.set_ylabel('Price ($)')
ax1.set_title(f'{symbol} Price and MACD')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(data.index, data['MACD'], label='MACD', linewidth=2)
ax2.plot(data.index, data['MACD_Signal'], label='Signal', linewidth=2)
ax2.bar(data.index, data['MACD_Diff'], label='Histogram', alpha=0.3)
ax2.set_ylabel('MACD')
ax2.set_xlabel('Date')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## Bollinger Bands

```python
# Calculate Bollinger Bands
bb = ta.volatility.BollingerBands(data['Close'], window=20, window_dev=2)
data['BB_Upper'] = bb.bollinger_hband()
data['BB_Middle'] = bb.bollinger_mavg()
data['BB_Lower'] = bb.bollinger_lband()

# Plot
plt.figure(figsize=(14, 8))
plt.plot(data.index, data['Close'], label='Close Price', linewidth=2)
plt.plot(data.index, data['BB_Upper'], label='Upper Band', alpha=0.5, linestyle='--')
plt.plot(data.index, data['BB_Middle'], label='Middle Band (SMA 20)', alpha=0.5)
plt.plot(data.index, data['BB_Lower'], label='Lower Band', alpha=0.5, linestyle='--')
plt.fill_between(data.index, data['BB_Upper'], data['BB_Lower'], alpha=0.1)
plt.title(f'{symbol} Price with Bollinger Bands')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
