# Getting Started with Trading AI

This notebook introduces the basics of algorithmic trading and data analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf

print("Libraries imported successfully!")
```

## Fetching Stock Data

```python
# Download stock data
symbol = "AAPL"
ticker = yf.Ticker(symbol)
data = ticker.history(period="1y")

print(f"Data shape: {data.shape}")
print(f"Columns: {data.columns.tolist()}")
data.head()
```

## Basic Price Visualization

```python
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Close'], label='Close Price')
plt.title(f'{symbol} Stock Price (1 Year)')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.show()
```

## Calculate Basic Statistics

```python
# Calculate returns
data['Returns'] = data['Close'].pct_change()

# Basic statistics
print("=== Price Statistics ===")
print(f"Current Price: ${data['Close'].iloc[-1]:.2f}")
print(f"52-Week High: ${data['Close'].max():.2f}")
print(f"52-Week Low: ${data['Close'].min():.2f}")
print(f"Average Daily Return: {data['Returns'].mean() * 100:.2f}%")
print(f"Volatility (Std Dev): {data['Returns'].std() * 100:.2f}%")
```
